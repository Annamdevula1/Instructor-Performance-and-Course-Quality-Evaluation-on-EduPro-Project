**Instructor Impact And Enrollment**

**Count Enrollments/Transactions per Instructor**

```python
enrollment_by_teacher = transactions.groupby(
    "TeacherID"
).size().reset_index(name="EnrollmentCount")
```

**Combine With Teacher Information**

```python
instructor_summary = teachers.merge(
    enrollment_by_teacher,
    on="TeacherID",
    how="left"
)

instructor_summary["EnrollmentCount"] = (
    instructor_summary["EnrollmentCount"].fillna(0)
)

display(instructor_summary.head())
```

**Teacher Rating VS Enrollment**

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=instructor_summary,
    x="TeacherRating",
    y="EnrollmentCount"
)

plt.title("Teacher Rating vs Enrollment Count")
plt.xlabel("Teacher Rating")
plt.ylabel("Enrollment Count")

plt.show()
```

```python
print(
    instructor_summary[
        ["TeacherRating", "EnrollmentCount"]
    ].corr()
)
```

**High/Medium/Low Instructor Performance**

**Create rating Tiers**

```python
instructor_summary["RatingTier"] = pd.cut(
    instructor_summary["TeacherRating"],
    bins=[0, 3, 4, 5],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)
```

**compare tiers**

```python
tier_analysis = instructor_summary.groupby(
    "RatingTier",
    observed=True
).agg(
    AverageTeacherRating=("TeacherRating", "mean"),
    AverageEnrollments=("EnrollmentCount", "mean"),
    NumberOfInstructors=("TeacherID", "count")
).reset_index()

display(tier_analysis)
```

**Tier Enrollment Graph**

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=tier_analysis,
    x="RatingTier",
    y="AverageEnrollments"
)

plt.title("Average Enrollment by Instructor Rating Tier")
plt.xlabel("Instructor Rating Tier")
plt.ylabel("Average Enrollment")

plt.show()
```

```python
# Missing KPI Calculations

# 1. Rating Consistency Index
rating_consistency_index = (
    1 - (df["TeacherRating"].std() / df["TeacherRating"].mean())
)

# 2. Experience Impact Score
experience_impact_score = df["YearsOfExperience"].corr(
    df["TeacherRating"]
)

# 3. Enrollment Influence Ratio
high_rated_enrollments = df.loc[
    df["TeacherRating"] >= 4,
    "TransactionID"
].nunique()

total_enrollments = df["TransactionID"].nunique()

enrollment_influence_ratio = (
    high_rated_enrollments / total_enrollments
)

# Display KPI values
print("Rating Consistency Index:", round(rating_consistency_index, 3))
print("Experience Impact Score:", round(experience_impact_score, 3))
print("Enrollment Influence Ratio:", round(enrollment_influence_ratio, 3))
```

```python
# KPI Summary Table

kpi_summary = pd.DataFrame({
    "KPI": [
        "Average Teacher Rating",
        "Average Course Rating",
        "Total Enrollments",
        "Total Instructors",
        "Total Courses"
    ],
    "Value": [
        df["TeacherRating"].mean(),
        df["CourseRating"].mean(),
        df["TransactionID"].nunique(),
        df["TeacherID"].nunique(),
        df["CourseID"].nunique()
    ]
})

display(kpi_summary)
```
